##  Setup — Conexão com o Banco

```python
import sqlite3
import pandas as pd

# Conexão com o banco
conn = sqlite3.connect('../olist.db')

# Configuração de exibição
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 20)

print('✅ Conexão estabelecida!')
```

##  Exploração do Schema

```python
# Listar todas as tabelas do banco
query = """
SELECT name 
FROM sqlite_master 
WHERE type = 'table'
ORDER BY name
"""

pd.read_sql_query(query, conn)
```

---
## 1. SELECT Básico

```python
# SELECT básico — primeiros registros da tabela

query = """
SELECT *
FROM tb_orders
LIMIT 10
"""

pd.read_sql_query(query, conn)
```

## 2. WHERE — Filtros


```python
# WHERE — Filtro por status do pedido

query = """
SELECT *
FROM tb_orders
WHERE order_status = 'delivered'
LIMIT 10
"""

pd.read_sql_query(query, conn)
```

## 3. ORDER BY — Ordenação

```python
# ORDER BY — Ordenação por data de compra

query = """
SELECT *
FROM tb_orders
ORDER BY order_purchase_timestamp DESC
LIMIT 10
"""

pd.read_sql_query(query, conn)
```

## 4. GROUP BY + Agregações

```python
# GROUP BY + COUNT — Pedidos por status
query = """
SELECT 
    order_status,
    COUNT(*) AS total_pedidos
FROM tb_orders
GROUP BY order_status
ORDER BY total_pedidos DESC
"""

pd.read_sql_query(query, conn)
```
