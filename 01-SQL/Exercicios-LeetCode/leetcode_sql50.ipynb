## Exercício 01 — Recyclable and Low Fat Products

**Dificuldade:** 🟢 Fácil  
**Tópico:** SELECT / WHERE / AND

> Encontrar os IDs de produtos que sejam tanto pobres em gordura quanto recicláveis.

-- Solução

```sql
SELECT product_id
FROM Products
WHERE low_fats = 'Y'
  AND recyclable = 'Y';
  ```

**Resultado esperado:** `product_id` → 1, 3

💡 **Nota:** filtro duplo com `AND` direto no `WHERE` — sem subquery necessária.

## Exercício 02 — Find Customer Referee

**Dificuldade:** 🟢 Fácil 

**Tópico:** SELECT / WHERE / IS NULL / OR

> Retornar nomes de clientes cujo `referee_id` é diferente de 2 ou é nulo.

-- Solução

```sql
SELECT name
FROM Customer
WHERE referee_id != 2
   OR referee_id IS NULL;
``` 

**Resultado esperado:** Will, Jane, Bill, Zack

💡 **Nota:** o ponto-chave aqui é que `NULL != 2` **não retorna TRUE** em SQL —  
comparações com `NULL` sempre resultam em `UNKNOWN`, por isso o `OR IS NULL` é obrigatório.  
Sem ele, clientes sem indicação (null) seriam silenciosamente excluídos do resultado.

## Exercício 03 — Big Countries

**Dificuldade:** 🟢 Fácil  
**Tópico:** SELECT / WHERE / OR

> Retornar nome, população e área de países considerados "grandes":  
> área ≥ 3.000.000 km² **ou** população ≥ 25.000.000.

-- Solução

```sql
SELECT name, population, area
FROM World
WHERE area >= 3000000
   OR population >= 25000000;
```

**Resultado esperado:** Afghanistan, Algeria

💡 **Nota:** a condição é `OR` — basta satisfazer **um** dos critérios para o país ser "grande".  
Usar `AND` aqui seria um erro lógico: exigiria que o país atendesse **ambos** ao mesmo tempo,  
reduzindo o resultado a apenas os países gigantes em área **e** população simultaneamente.
