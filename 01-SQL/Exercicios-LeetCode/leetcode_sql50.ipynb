## Exercício 01 — Recyclable and Low Fat Products

**Dificuldade:** 🟢 Fácil  
**Tópico:** SELECT / WHERE / AND

> Encontrar os IDs de produtos que sejam tanto pobres em gordura quanto recicláveis.

-- Solução

```sql
SELECT product_id
FROM Products
WHERE low_fats = 'Y'
  AND recyclable = 'Y';
  ```

**Resultado esperado:** `product_id` → 1, 3

💡 **Nota:** filtro duplo com `AND` direto no `WHERE` — sem subquery necessária.

---

## Exercício 02 — Find Customer Referee

**Dificuldade:** 🟢 Fácil 

**Tópico:** SELECT / WHERE / IS NULL / OR

> Retornar nomes de clientes cujo `referee_id` é diferente de 2 ou é nulo.

-- Solução

```sql
SELECT name
FROM Customer
WHERE referee_id != 2
   OR referee_id IS NULL;
``` 

**Resultado esperado:** Will, Jane, Bill, Zack

💡 **Nota:** o ponto-chave aqui é que `NULL != 2` **não retorna TRUE** em SQL —  
comparações com `NULL` sempre resultam em `UNKNOWN`, por isso o `OR IS NULL` é obrigatório.  
Sem ele, clientes sem indicação (null) seriam silenciosamente excluídos do resultado.

---

## Exercício 03 — Big Countries

**Dificuldade:** 🟢 Fácil  
**Tópico:** SELECT / WHERE / OR

> Retornar nome, população e área de países considerados "grandes":  
> área ≥ 3.000.000 km² **ou** população ≥ 25.000.000.

-- Solução

```sql
SELECT name, population, area
FROM World
WHERE area >= 3000000
   OR population >= 25000000;
```

**Resultado esperado:** Afghanistan, Algeria

💡 **Nota:** a condição é `OR` — basta satisfazer **um** dos critérios para o país ser "grande".  
Usar `AND` aqui seria um erro lógico: exigiria que o país atendesse **ambos** ao mesmo tempo,  
reduzindo o resultado a apenas os países gigantes em área **e** população simultaneamente.

## Exercício 04 — Article Views I

**Dificuldade:** 🟢 Fácil  
**Tópico:** SELECT / WHERE / DISTINCT / ORDER BY

> Retornar o id de todos os autores que visualizaram pelo menos um de seus próprios:  
> artigos: `author_id` = `viewer_id`, ordenado de forma crescente.

-- Solução

```sql
SELECT DISTINCT author_id AS id
FROM views
WHERE author_id = viewer_id
ORDER BY id ASC;
```

**Resultado esperado:** `id = 4, 7`
💡 **Nota:** o `DISTINCT` é essencial aqui — a tabela pode ter o mesmo autor aparecendo múltiplas vezes (como o `author_id` = 4 que aparece duas vezes no exemplo). Sem ele, o resultado traria duplicatas. O `AS id` renomeia a coluna para exatamente o nome que o enunciado exige — sem isso, a resposta seria considerada errada mesmo com os dados corretos.

## Exercício 05 — Invalid Tweets

**Dificuldade:** 🟢 Fácil  
**Tópico:** SELECT / WHERE / LENGTH

> Retornar o `tweet_id` de todos os tweets cujo conteúdo tenha mais de 15 caracteres:
> `LENGTH(content) > 15`

-- Solução

```sql
SELECT tweet_id
FROM tweets
WHERE LENGTH(content) > 15;
```

**Resultado esperado:**  todos os `tweet_id` cujo conteúdo ultrapasse 15 caracteres. 

💡 **Nota:** `LENGTH()` conta os caracteres da string. A condição é estritamente maior que 15 — tweets com exatamente 15 caracteres são válidos e ficam fora do resultado. Usar >= 15 seria um erro lógico: incluiria tweets válidos no resultado.

## Exercício 06 — Replace Employee ID With The Unique Identifier

**Dificuldade:** 🟢 Fácil  
**Tópico:** SELECT / LEFT JOIN / NULL

> Retornar o `unique_id` de cada usuário. Se o usuário não possuir um ID único, exibir `null`:
> `LEFT JOIN` garante que todos os funcionários apareçam, mesmo sem correspondência em `EmployeeUNI`



-- Solução

```mysql
SELECT uni.unique_id, e.name
FROM Employees e
LEFT JOIN EmployeeUNI uni
ON e.id = uni.id;
```

**Resultado esperado:** todos os funcionários listados com seu `unique_id`. Funcionários sem correspondência em `EmployeeUNI` aparecem com `null` no lugar do ID único.

💡 **Nota:** `LEFT JOIN` é essencial aqui — um `INNER JOIN` excluiria silenciosamente os funcionários sem `unique_id`, o que violaria o enunciado. O `null` não é ausência de resultado, é o resultado correto para esses casos.

## Exercício 07 — Sales Analysis

**Dificuldade:** 🟢 Fácil  
**Tópico:** SELECT / JOIN

> Retornar o `product_name`, `year` e `price` para cada `sale_id` na tabela `Sales`:
> `JOIN` com a tabela `Product` para obter o nome do produto a partir do `product_id`


-- Solução

```mysql
SELECT 
product_name,
year,
price
FROM Product
INNER JOIN Sales ON Sales.product_id = Product.product_id;

```

**Resultado esperado:** todas as vendas listadas com o nome do produto correspondente, o ano e o preço.

💡 **Nota:** `JOIN` (inner) é suficiente aqui — todo `sale_id` possui um `product_id` válido na tabela `Product`, então nenhuma linha seria perdida com um `INNER JOIN`. Se houvesse vendas com `product_id` sem correspondência, um `LEFT JOIN` seria necessário para preservá-las.

## Exercício 08 — Customer Who Visited but Did Not Make Any Transactions

**Dificuldade:** 🟢 Fácil  
**Tópico:** SELECT / LEFT JOIN / NULL / GROUP BY

> Retornar o `customer_id` e a quantidade de visitas sem transação (`count_no_trans`):
> `LEFT JOIN` com `Transactions` expõe visitas sem correspondência — onde `transaction_id IS NULL`, a visita não gerou nenhuma compra


-- Solução

```mysql

SELECT
    v.customer_id,
    COUNT(*) AS count_no_trans
FROM Visits v
LEFT JOIN Transactions t ON v.visit_id = t.visit_id
WHERE t.transaction_id IS NULL
GROUP BY v.customer_id;
```

**Resultado esperado:** todos os clientes que visitaram sem realizar nenhuma transação, com a contagem de quantas vezes isso ocorreu.

💡 **Nota:** `LEFT JOIN` é essencial aqui — ele mantém todas as visitas, inclusive as sem transação. Filtrar por `transaction_id IS NULL` isola exatamente as visitas sem compra. Um `INNER JOIN` excluiria silenciosamente essas visitas, zerando o resultado correto.

## Exercício 09  — Rising Temperature

**Dificuldade:** 🟢 Fácil  
**Tópico:** SELECT / SELF JOIN / DATE_SUB

>Encontrar todos os `ids` cujas temperaturas foram mais altas em comparação ao dia anterior: usar SELF JOIN na tabela `Weather` combinando `DATE_SUB` para encontrar o dia anterior — onde `w1.temperature` > ` w2.temperature`, a temperatura subiu

-- Solução

```mysql
SELECT w1.id
FROM Weather w1
JOIN Weather w2
  ON w2.recordDate = DATE_SUB(w1.recordDate, INTERVAL 1 DAY)
WHERE w1.temperature > w2.temperature;
```

**Resultado esperado:** todos os dias em que a temperatura foi maior que a do dia anterior, retornando apenas o id correspondente.

💡 **Nota:** O `SELF JOIN` é essencial aqui — ele une a tabela com ela mesma para comparar linhas distintas. Usar `DATE_SUB``(..., INTERVAL 1 DAY)` garante que apenas dias consecutivos sejam comparados. Um `LEFT JOIN` incluiria dias sem par anterior, quebrando a comparação.

## Exercício 10 — Average Time of Process per Machine

**Dificuldade:** 🟢 Fácil  
**Tópico:** SELECT / SELF JOIN / AVG / ROUND

> Encontrar o tempo médio que cada máquina leva para concluir um processo:
> usar SELF JOIN na tabela `Activity` combinando `machine_id` e `process_id`
> para parear cada evento `start` com seu respectivo `end` — calculando
> `AVG(end.timestamp - start.timestamp)` agrupado por `machine_id`

-- Solução

```mysql
SELECT 
    a.machine_id,
    ROUND(AVG(b.timestamp - a.timestamp), 3) AS processing_time
FROM Activity a
JOIN Activity b
    ON a.machine_id = b.machine_id
    AND a.process_id = b.process_id
    AND a.activity_type = 'start'
    AND b.activity_type = 'end'
GROUP BY a.machine_id;
```

**Resultado esperado:** todos os dias em que a temperatura foi maior que a do 
dia anterior, retornando apenas o `id` correspondente.

💡 **Nota:** O `SELF JOIN` é essencial aqui — ele une a tabela com ela mesma 
para comparar linhas distintas. `DATE_SUB(..., INTERVAL 1 DAY)` garante que 
apenas dias consecutivos sejam comparados. Um `LEFT JOIN` incluiria dias sem 
par anterior, quebrando a lógica de comparação.

## Exercício 11 — Employee Bonus

**Dificuldade:** 🟢 Fácil  
**Tópico:** SELECT / LEFT JOIN / WHERE / IS NULL

> Encontrar o nome e o valor do bônus de cada funcionário que possui bônus
> menor que `1000` ou não recebeu nenhum bônus:
> usar LEFT JOIN entre `Employee` e `Bonus` pelo `empId` para preservar
> funcionários sem registro de bônus — filtrando com
> `WHERE bonus < 1000 OR bonus IS NULL`

-- Solução

```mysql

SELECT e.name, b.bonus
FROM Employee e
LEFT JOIN Bonus b ON e.empId = b.empId
WHERE b.bonus < 1000 OR b.bonus IS NULL;


```

**Resultado esperado:** todos os funcionários cujo bônus é menor que `1000` 
ou que não receberam nenhum bônus, retornando `name` e `bonus` correspondentes.

💡 **Nota:** O `LEFT JOIN` é essencial aqui — ele preserva todos os funcionários
da tabela `Employee`, mesmo os que não possuem registro em `Bonus`, resultando
em `NULL` nesses casos. Um `INNER JOIN` excluiria esses funcionários, quebrando
a lógica de capturar quem não recebeu bônus.
