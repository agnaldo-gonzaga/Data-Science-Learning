```python
import pandas as pd
import numpy as numpy
```

```python
# Dicionário com dados fictícios de vendas: vendedor, região, produto, trimestre, vendas, meta e unidades

dados = {
    "vendedor":   ["Ana", "Bruno", "Ana", "Carlos", "Bruno", "Carlos", "Ana", "Bruno", "Carlos", "Ana"],
    "regiao":     ["Sul", "Norte", "Norte", "Sul", "Sul", "Norte", "Leste", "Leste", "Leste", "Sul"],
    "produto":    ["Notebook", "Tablet", "Notebook", "Tablet", "Smartphone", "Notebook", "Tablet", "Notebook", "Smartphone", "Smartphone"],
    "trimestre":  ["Q1", "Q1", "Q2", "Q1", "Q2", "Q2", "Q1", "Q2", "Q1", "Q2"],
    "vendas":     [12000, 8500, 15000, 9300, 11200, 7800, 6400, 13000, 10500, 9800],
    "meta":       [10000, 9000, 13000, 8000, 10000, 8000, 7000, 12000, 9000, 9000],
    "unidades":   [4, 17, 5, 18, 8, 6, 13, 4, 7, 3]
}
```

```python
df = pd.DataFrame(dados)
```

```python
df.head(10)
```

```python
# Pivot table:  vendas por vendedor e trimestre

pivot_df = df.pivot_table(index="vendedor", columns="trimestre", values="vendas", aggfunc="sum")
pivot_df
```

```python
# Pivot table: total de vendas por vendedor

pivot_total_venda = df.pivot_table(index="vendedor", columns="trimestre", values="vendas", aggfunc="sum", margins=True, margins_name="Total")
pivot_total_venda
```

```python
# Pivot table: Multiplas funções de agregação

multi_pivot = df.pivot_table(index="vendedor", columns="trimestre", values="vendas", aggfunc=["sum", "mean", "max"])
multi_pivot
```
