## 📌 Sobre o Dataset
O Iris Dataset é um dos datasets mais clássicos do ML, contendo **150 amostras** de flores divididas em **3 espécies**.

## 🗂️ Estrutura dos Dados
| Atributo | Descrição |
|---|---|
| `sepal length` | Comprimento da sépala (cm) |
| `sepal width` | Largura da sépala (cm) |
| `petal length` | Comprimento da pétala (cm) |
| `petal width` | Largura da pétala (cm) |

## 🎯 Classes (target)
- `0` → Iris Setosa
- `1` → Iris Versicolor
- `2` → Iris Virginica

---

### Importação dos dados

````python
from sklearn.datasets import load_iris
iris = load_iris()
````

### Observaçõs

````python
x = iris.data
print(x)
````

### Target

````python
y = iris.target # para mostrar o nome target_name
print(y) 
````

#### Shape das obrsações

````python
print(iris.data.shape)
````

````python
print(iris.target.shape)
````

###  Importação e Configuração do KNN (K-Nearest Neighbors)
> Algoritmo de classificação baseado nos **K vizinhos mais próximos**
---

````python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=1) # Instancia o modelo KNN com K=1 (classifica com base no vizinho mais próximo)
````

### Treinar a máquina

````python
knn.fit (x, y)
````

### Fazer Previsões

````python
species = knn.predict([[5.9, 3.0, 5.1, 1.8]])[0] # Realiza a predição com base na amostra [sepal_length, sepal_width, petal_length, petal_width]
print(iris.target_names[species])
````

### Separar os dados em dois grupos

````python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.25)
print(x_train.shape)
print(x_test.shape)
````

### Avaliação da performance do modelo

````python
knn.fit (x_train,y_train)
previsoes = knn.predict(x_test)
````

````python
from sklearn import metrics
acertos = metrics.accuracy_score(y_test, previsoes)
print(acertos)  
````

###  Testando diferentes valores de K

> Avalia a acurácia do KNN para K entre 1 e 25, armazenando os resultados em um dicionário


````python
valores_performance = {}
k = 1

while k <= 25:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(x_train, y_train)
    previsoes = knn.predict(x_test)
    acertos = metrics.accuracy_score(y_test, previsoes)
    valores_performance[k] = round(acertos, 4)
    k += 1
print(valores_performance)
````

````python
import matplotlib.pyplot as plt

plt.plot(list(valores_performance.keys()), list(valores_performance.values()))
plt.xlabel('Valor de K')
plt.ylabel('Performance (Acurácia)')
plt.title('Performance do KNN para diferentes valores de K')
plt.show()
````

###  Interpretando o Gráfico — Performance do KNN por valor de K

> O gráfico mostra como a **acurácia do modelo muda** conforme aumentamos o valor de K

---

#### 🔍 O que o gráfico representa
- **Eixo X** → Valor de K (número de vizinhos considerados)
- **Eixo Y** → Acurácia do modelo (quanto maior, melhor)

---

#### 📌 Conclusões visuais

| Região | O que acontece |
|---|---|
| **K = 1 a 4** | ✅ Acurácia alta (~0.973) |
| **K = 8 a 9** | ⚠️ Pior desempenho (~0.921) |
| **K = 11 a 20** | ✅ Acurácia estável (~0.973) |
| **K = 22 a 25** | 📉 Leve queda (~0.947) |

---

#### 🧠 Resumo simples
- **Nem sempre K maior = melhor resultado**
- O modelo teve **quedas bruscas** em K=8 e K=9
- Os melhores valores de K para esse dataset ficaram entre **1-4** e **11-20**

### 🌳 Decision Tree — Árvore de Decisão

> Algoritmo que aprende **regras de decisão** a partir dos dados, como um fluxograma de perguntas

---

#### 🧠 Como funciona
- Divide os dados em grupos fazendo perguntas sobre as features
- Exemplo: *"O comprimento da pétala é menor que 2.5cm?"* → Se sim, é **Setosa**
- Vai criando divisões até conseguir classificar corretamente

````python
from sklearn.tree import DecisionTreeClassifier
# Instancia o modelo com configurações padrão (sem limitar profundidade da árvore)
dt = DecisionTreeClassifier()
dt.fit(x_train, y_train)
````

````python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt

plt.figure(figsize=(16, 8))
plot_tree(
    dt,
    feature_names=iris.feature_names,  # nomes reais das colunas
    class_names=iris.target_names,      # nomes reais das espécies
    filled=True,                        # colorido por classe
    rounded=True                        # bordas arredondadas
)
plt.title('Árvore de Decisão — Iris Dataset')
plt.show()
````

---

### 🌳 Entendendo o Gráfico — Árvore de Decisão

> A árvore aprende **fazendo perguntas** sobre as features até conseguir classificar cada flor

---

#### 🔍 Como ler cada caixa
| Campo | O que significa |
|---|---|
| **Pergunta no topo** | A regra que o modelo aprendeu (ex: `petal width <= 0.8`) |
| **gini** | Impureza do grupo — `0.0` = perfeito, todos da mesma classe |
| **samples** | Quantas amostras chegaram até esse ponto |
| **value** | Quantidade por classe `[Setosa, Versicolor, Virginica]` |
| **class** | Classe majoritária nesse nó |

---

#### 🧠 O que esse modelo aprendeu
- **Setosa** → separada logo na 1ª pergunta (`petal width <= 0.8`) — pétalas menores
- **Versicolor vs Virginica** → precisam de mais perguntas para separar
- Nós com `gini = 0.0` → grupo puro, classificação certeira
- Nós com `samples = 1` → ⚠️ criou regra pra 1 amostra só — sinal de **overfitting**

---

#### ⚠️ Overfitting — o que é?
O modelo decorou os dados de treino em vez de aprender padrões gerais.
Solução → limitar a profundidade da árvore:
```python
dt = DecisionTreeClassifier(max_depth=3)  # árvore mais simples e generalizável
```
